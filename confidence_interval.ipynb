```python

import scipy.stats as sp
import matplotlib.pyplot as plt
import numpy as np
import math
```

```python
#50 students , mean =165,sd=10;95%CI(1.96)
mean=165
sd=10
n =50
z=1.96
margin_error=z*(sd/math.sqrt(n))
lower = mean-margin_error
upper =mean+margin_error
print("95%confidence interval")
print(lower,upper)


```
