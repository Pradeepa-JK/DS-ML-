```python
import pandas as pd
import numpy as np
import scipy.stats as st
import math

```

```python

dt= pd.read_csv("/Titanic-Dataset.csv")
age = dt["Age"].dropna()

confidence_interval = st.t.interval(confidence=0.95,df=len(age)-1,loc=np.mean(age),scale=st.sem(age))
print("95% Confidence Interval:")
print(confidence_interval)
```
