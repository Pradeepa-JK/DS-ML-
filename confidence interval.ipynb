```python
import scipy.stats as st
import math
```

```python
sample_mean = 50
std_dev = 10
sample_size = 100
confidence = 0.95
std_error = std_dev / math.sqrt(sample_size)
interval = st.norm.interval(
    confidence,
    loc=sample_mean,
    scale=std_error
)

print("Confidence Interval:")
print(interval)
```
