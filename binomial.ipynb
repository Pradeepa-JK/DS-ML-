```python
from scipy.stats import binom
import numpy as np
import matplotlib.pyplot as plt
```

```python
#95% defect free items ,probability exactly 9 out of 10
p=0.95
n=10
k=9

prob1=binom.pmf(k,n,p)
print("probability ",prob1)
```

```python
x=np.arange(0,n+1)
pmf_values=binom.pmf(x,n,p)
plt.bar(x,pmf_values,color="green",edgecolor="black",label="Binomial PMF")
plt.bar(k,prob1,color="red",edgecolor="black",label=f"k = {k}")
plt.xlabel("Number of Exactly good ones")
plt.ylabel("Probability")
plt.title(f"Binomial distribution")
plt.legend()
plt.grid(axis="y",alpha=0.6)
plt.show()
```

```python
#student solves 60% problems at least 3 of 5

n=5
p=0.6
probability=0
for k in range(3,6):
    probability+=binom.pmf(k,n,p)
print(probability)
```

```python
x = np.arange(0, n + 1)
pmf_values = binom.pmf(x, n, p)
k_values = np.arange(3, 6)
probability = binom.pmf(k_values, n, p)
plt.bar(x, pmf_values,
        color="violet",
        edgecolor="black",
        label="Binomial PMF")
plt.bar(k_values, probability,
        color="skyblue",
        edgecolor="black",
        label="k ≥ 3")

plt.xlabel("Number of Success")
plt.ylabel("Probability")
plt.title("Binomial Distribution")
plt.legend()
plt.grid(axis="y", alpha=0.5)
plt.show()
```

```python
p = 0.7
n = 8
k = 6
prob3 = binom.pmf(k, n, p)
print("Probability of making exactly 6 shots:", prob3)
x = np.arange(0, n + 1)
# PMF values
pmf_values = binom.pmf(x, n, p)
```
