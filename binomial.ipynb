```python
import matplotlib.pyplot as plt
from scipy.stats import binom
import numpy as np 
```

```python
'''
factory produces bulbs.
The probability that a bulb is defective is 0.1.

If 8 bulbs are checked, find the probability that exactly 1 bulb is defective.
no of trails(n)=8
no of  defective(k)=1
probability of defective (p)=0.1
'''
n=8
p=0.1
k=1
result=binom.pmf(k,n,p)
print(result)
```

```python
x=np.arange(0,n+1)
pmf_values=binom.pmf(x,n,p)
plt.bar(x,pmf_values,color="blue",edgecolor="black",label="binomial PMF")
plt.bar(k,result,color="green",edgecolor="black",label=f"k={k}")
plt.xlabel("number of defective bulbs")
plt.ylabel("probability")
plt.title("Binomial distribution")
plt.legend()
plt.show()
```
