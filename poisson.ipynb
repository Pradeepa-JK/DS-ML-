```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson


```

```python
'''
A traffic signal receives an average of 4 cars per minute.
Find the probability that exactly 6 cars arrive in one minute.
'''

lam = 4
k = 6
result = poisson.pmf(k, lam)
print(result)
```

```python
x = np.arange(0, 15)
pmf_values = poisson.pmf(x, lam)
plt.bar(x, pmf_values,color="lightgreen",edgecolor="black",label="Poisson PMF")
plt.bar(k, result,color="orange",edgecolor="black",label=f"k={k}")
plt.xlabel("Number of Cars")
plt.ylabel("Probability")
plt.title("Poisson Distribution")
plt.legend()
plt.grid(axis="y", alpha=0.6)

plt.show()
```

```python
'''
A hospital receives an average of 2 emergency cases per hour.
Find the probability that exactly 4 emergency cases arrive in one hour.
'''
lam = 2
k = 4
result = poisson.pmf(k, lam)
print(result)
x = np.arange(0, 12)
```

```python
pmf_values = poisson.pmf(x, lam)
plt.bar(x, pmf_values,color="lightblue",edgecolor="black",label="Poisson PMF")
plt.bar(k, result,color="green",edgecolor="black",label=f"k={k}")
plt.xlabel("Number of Emergency Cases")
plt.ylabel("Probability")
plt.title("Poisson Distribution")
plt.legend()
plt.grid(axis="y", alpha=0.6)
plt.show()
```
