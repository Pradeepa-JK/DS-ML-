```python
from scipy.stats import poisson
import numpy as np
import matplotlib.pyplot as plt
```

```python
lam = 6
k = 10
prob1 = poisson.pmf(k, lam)
print("Probability of exactly 10 visits:", prob1)
```

```python
x = np.arange(0, 16)
pmf_values = poisson.pmf(x, lam)
plt.figure(figsize=(6,4))

plt.bar(x, pmf_values,color="lightblue",edgecolor="black",label="Poisson PMF")
plt.bar(k, prob1,color="red",edgecolor="black",label=f"k = {k}")
plt.xlabel("Number of Visits")
plt.ylabel("Probability")
plt.title("Question 1 - Poisson Distribution")
plt.legend()
plt.grid(axis="y", alpha=0.6)
plt.show()
```
